# مرحلة معالجة البيانات وتنضيفها

```python
df = df[["comment", "sentiment"]]
df = df.dropna()
```

```python
df = df[df["sentiment"].isin(["Positive", "Negative"])]
```

```python
df["sentiment"].value_counts()
```

```python
df["label"] = df["sentiment"].map({
    "Negative": 0,
    "Positive": 1
})
```

```python
df[["sentiment","label"]].head()
df["label"].unique()
```

```python
df=df.dropna()
```

```python
df.isna().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df["comment"],df["sentiment"])
```

```python
import string
string.punctuation
```

```python
pun_comment=str.maketrans("","",string.punctuation)
```

```python
df["comment"]=df["comment"].str.translate(pun_comment)
```

```python
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["comment"],
    df["label"],
    test_size=0.2,ييي
    random_state=42,
    stratify=df["label"]
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1,2)
)

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

model.fit(X_train_vec, y_train)
```

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_vec)

print(classification_report(y_test, y_pred))
```

```python
print(df["label"].unique())
print(df["label"].dtype)
```

```python
def predict_comment(text):
    vec = vectorizer.transform([text])
    pred = model.predict(vec)[0]
    return pred
```

```python
print(predict_comment("المقال جيد لكن طويل"))
print(predict_comment(" انت حمار  "))
print(predict_comment("طرح لا بأس به"))
print(predict_comment("الصراحة توقعت أفضل"))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import arabic_reshaper
from bidi.algorithm import get_display
```

```python
def format_arabic_text(text):
    if isinstance(text, str):
        reshaped_text = arabic_reshaper.reshape(text)
        bidi_text = get_display(reshaped_text)
        return bidi_text
    return text
```

# مرحلة تدريب النموذج

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.svm import LinearSVC
```

```python
x_cpmment=df["comment_tokinize"]
y_sarcasm=df["sentiment"]
```

```python
x_train,x_test,y_train,y_test =train_test_split(x_cpmment,y_sarcasm,test_size=0.2,random_state=42)
```

```python
commet_classfiction=Pipeline([("tfd",TfidfVectorizer()),("svm",LinearSVC())])
```

```python
commet_classfiction.fit(x_train,y_train)
```

```python
commet_predection=commet_classfiction.predict(x_test)
```

```python
accuracy=accuracy_score(commet_predection,y_test)
```
