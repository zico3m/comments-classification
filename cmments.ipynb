# مرحلة معالجة البيانات وتنضيفها

```python
df=df.dropna()
```

```python
df.isna().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df["comment"],df["title"])
```

```python
import string
string.punctuation
```

```python
pun_comment=str.maketrans("","",string.punctuation)
```

```python
df["comment"]=df["comment"].str.translate(pun_comment)
```

```python
df.sample(5)
```

```python
import nltk 
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
df["comment_tokinize"]=df["comment"].apply(word_tokenize)
```

```python
table = df[["comment", "comment_tokinize"]].sample(10)
```

```python
table.sample(5)
```

```python
stopwords=set(stopwords.words("arabic"))
```

```python
stopwords
```

```python
df["comment_tokinize"]=df["comment_tokinize"].apply(lambda x :[ word for word in x if word not in stopwords ])
```

```python
table.sample(5)
```

```python
df["comment_tokinize"]=df["comment_tokinize"].apply(lambda x : ' '.join(x))
```

```python
table.sample(5)
```

```python
from sklearn.preprocessing import LabelEncoder
eccoder= LabelEncoder()
```

```python
df["dangerous"]=eccoder.fit_transform(df["dangerous"])
```

# استكشاف البيانات بعد المعالجة 

```python
import matplotlib.pyplot as plt
import seaborn as sns
import arabic_reshaper
from bidi.algorithm import get_display
```

```python
def format_arabic_text(text):
    if isinstance(text, str):
        reshaped_text = arabic_reshaper.reshape(text) 
        bidi_text = get_display(reshaped_text)         
        return bidi_text
    return text
```

```python
sentiment_counts = df["sentiment"].value_counts()
plt1 =plt.figure(figsize=(6,4))
sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values, palette="viridis")
plt.title(format_arabic_text("توزيع المشاعر في التعليقات"), fontsize=16)
plt.xlabel(format_arabic_text(("نوع المشاعر")), fontsize=12)
plt.ylabel(format_arabic_text("عدد التعليقات"), fontsize=12)
plt.show()
```

# مرحلة تدريب النموذج

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.svm import LinearSVC
```

```python
x_cpmment=df["comment_tokinize"]
y_sarcasm=df["sentiment"]
```

```python
x_train,x_test,y_train,y_test =train_test_split(x_cpmment,y_sarcasm,test_size=0.2,random_state=42)
```

```python
commet_classfiction=Pipeline([("tfd",TfidfVectorizer()),("svm",LinearSVC())])
```

```python
commet_classfiction.fit(x_train,y_train)
```

```python
commet_predection=commet_classfiction.predict(x_test)
```

```python
accuracy=accuracy_score(commet_predection,y_test)
```
